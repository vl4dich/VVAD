Рассмотрим величину $m$ - вероятность верной классификации, тогда требуемая величина $q$ - вероятность ошибочной классификации - в точности равна
$$
q = 1 - m.
$$

В предположении, что распределение вероятностей в train и в test одинаково, получаем, что вероятности классов в обеих выборках равны
$$
p_1, p_2, \dots, p_K, \qquad \sum_{i=1}^{K} p_i = 1.
$$

### 1) Случайный классификатор по частотам train

Найдем вероятность того, что мы классифицировали объект правильно в случае классификатора, который выдает случайный класс в соответствии с частотой встречаемости классов в train-выборке. То есть независимо от объекта это случайная величина $\xi$ с вероятностями
$$
\mathbb{P}(\xi = i) = p_i,\qquad i=1,\dots,K.
$$

Пусть $\mathbb{P}(\xi=i \mid \text{object}=i)$ - вероятность правильно назвать объект $i$-го типа. Тогда
$$
\mathbb{P}(\xi=i \mid \text{object}=i)=\mathbb{P}(\xi=i)=p_i
\quad \text{(из-за независимости)}.
$$

Распишем полную вероятность правильно классифицировать (запишем формулу полной вероятности):
$$
m_1=\sum_{i=1}^{K}\mathbb{P}(\xi=i \mid \text{object}=i)\,\mathbb{P}(\text{object}=i)
=\sum_{i=1}^{K} p_i\cdot p_i
=\sum_{i=1}^{K} p_i^2.
$$

### 2) Классификатор, всегда выдающий самый частый класс

Найдем вероятность того, что мы классифицировали объект правильно в случае классификатора, который всегда выдает самый частый класс в train-выборке.

Пусть $\mathbb{P}(\xi=i \mid \text{object}=i)$ - вероятность правильно назвать объект $i$-го типа. Пусть $j$ - самый частый класс в train-выборке, т.е.
$$
\forall\, i \in \{1,\dots,K\}\quad p_j \ge p_i .
$$

Тогда
$$
\mathbb{P}(\xi=i \mid \text{object}=i)=
\begin{cases}
1, & i=j,\\
0, & \text{иначе}.
\end{cases}
$$

Теперь распишем полную вероятность:
$$
m_2=\sum_{i=1}^{K}\mathbb{P}(\xi=i \mid \text{object}=i)\,\mathbb{P}(\text{object}=i)
=p_j.
$$

### 3) Сравнение качеств: $m_2 \ge m_1$

Теперь докажем, что $m_2 \ge m_1$:
$$
m_2=p_j
= p_j \cdot \sum_{i=1}^{K} p_i
\ge \sum_{i=1}^{K} p_i\cdot p_i
=\sum_{i=1}^{K} p_i^2
=m_1.
$$

Равенство достигается только в тривиальном случае, когда $p_j=1$ или $p_i=p_j$ для любых $i$ и $j$.

### `meshgrid `
 функция получает на вход n одномерных массивов (каждый массив соответствует какой-то координате), на выходе дает n, n мерных массивов, где в ячейках i-ого массива находятся только i-ая координата. То есть функция растягивает каждый массив до размерности n.
Аргументы:
* `indexing` - определяет порядок координат
* `sparce` - определяет будет возвращен полноценный массив или разреженный, который готов растянуться через broadcasting
* `copy` - гарантия, что результат отдельный масси, а не новый заголовок

### `Figure`
 основной контейнер `plotly`, в нём хранится:
* `data` - данные
* `layout` - оформление графика
* `frames` - кадры (если график с анимацией)

### `Surface`
 слой данных (data), который выводится как поверхность
* `z` - двумерный массив, ячейки которого - значения, которые будут на графике
* `x`, `y` - массив с координатами узлов по x и по y
* `colorscale` - непосредственно выбор цвета
* `showscale` - показывать ли цветовую шкалу

### `update_layout`
 метод, который обновляет оформление графика (заголовок, размер, отступы, подписи осей, легенда и т.д.)
 
### `scene`
 опция в layout, которая отвечает за оформление 3D графика:
* `*axis_title` - настройка заголовка (вместо * ставим `x`/`y`/`z`)
* `aspectmode` - точка наблюдателя (`"cube"`, `"data"`, `"auto"`, ...)
* другие аргументы (сетка, диапазон, ручная настройка положения камеры)

### `margin`
 отступ между краем всего рисунка и содержимого графика
* `l`/`r`/`d`/`t` - отступ слева/справа/снизу/сверху в пикселях

#### 2. Классификация логическими правилами

На основе выше проведенного исследования предложите два разных правила вида
> 1). *если признак $x_1 > 5$, то класс 0, иначе посмотрим на признак $x_2$ и ...*
> 
> 2). *если $x_1 > 5$ и $x_2 < 2$, то класс 1, иначе ...*

В совокупности используйте не менее трех различных признаков. Никакие статистические методы использовать не требуется, только логику и графики. Любые решения "из коробки" использовать запрещено. Например, достаточно пояснения "*по графику <описание графика> визуально я увидел, что при X > 100 большинство пассажиров выжило, поэтому использую правило <описание правила>*".


Сделаем одно, но большое правило, что разрешено (исходя из чата курса)

Во-первых я увидел, что самое большое различие по признаку пола, поэтому далее рассматривал остальные признаки в зависимости от него. 
Женщины, внезависимости от комбинации класса или возраста чаще выживают. Мужчины, которые не в первом классе, почти не выживают, среди мужчин чаще выживают, чем не выживают только молодые (до 15 лет) или с дорогим билетом (>120)


**Правило:**

if `Sex` = `'female'`:
- `Survived` = `1`

else `Sex` = `'male'`:
- if `Pclass` != `1`:
-  - `Survived` = `0`
- elif `Age` < `15` or `Fare` > `120`:
- - `Survived` = `1`


Реализуйте полученные логические правила по шаблону

```python
def model(features: pd.DataFrame) -> pd.Series:
    """
    Предсказание целевого признака для данных features -- pandas-таблица данных.
    Возвращает pandas.Series с теми же индексами, что и у features.
    """

    predicted = pd.Series(0, index=features.index, dtype=int)

    predicted.loc[features['Sex'] == 'female'] = 1

    mask = ( 
        (features['Sex'] == 'male')
        & (features['Pclass'] == 1)
        & ((features['Age'] < 15)
        | (features['Fare'] >= 120))
    )

    predicted.loc[mask] = 1
    return predicted
```

Проверьте работоспособность данных моделей на тестовой выборке и оцените качество модели по метрике `accuracy`.

> &#x2755; *Примечание.* Полученные значения не влияют на оценку по заданию. Оценивается только корректность и обоснованность решения, а также графики и выводы.

```python
def accuracy(target: pd.Series, predicted: pd.Series) -> float:
    """
    Вычисление критерия качества для предсказания predicted,
    если истинные значения -- target.
    Возвращает одно вещественное число.
    """
    return (target == predicted).mean()
```

```python
print(f"Точность равна: {np.round(accuracy(y_train, model(X_train)) * 100,
                                  decimals=2)} %")
```

```python
mask_train = X_train.notna().all(axis=1)
mask_test = X_test.notna().all(axis=1)

X_train = X_train.loc[mask_train].copy()
y_train = y_train.loc[mask_train].copy()
X_test = X_test.loc[mask_test].copy()
y_test = y_test.loc[mask_test].copy()
```

Постройте классификатор методом ближайшего соседа. Обучите его и оцените качество модели. Сравните с предыдущим методом.

Пока никаких других преобразований данных делать не нужно.

```python
def knn(X_train: pd.DataFrame, y_train: pd.Series, X_test: pd.DataFrame, k: int) -> pd.Series:
    predict = []
    Xtr = X_train.to_numpy()
    ytr = y_train.to_numpy()
    Xte = X_test.to_numpy()
    for i in range(Xte.shape[0]):
        x = Xte[i]
        dists = np.sqrt(((Xtr - x) ** 2).sum(axis=1))
        idx = np.argsort(dists)[:k] # Сортируем расстояния "в лоб"
        vals, cnts = np.unique(ytr[idx], return_counts=True)
        predict.append(vals[np.argmax(cnts)])

    return pd.Series(predict, index=X_test.index)
```

```python
pred = knn(X_train, y_train, X_test, 5)

print(f"Точность равна: {np.round(accuracy(y_test, pred) * 100, decimals=2)}%")
```

Для решения этой проблемы нужно привести признаки к единому масштабу. Существует много методов масштабирования рассмотрим работу одного из них &mdash; **минимаксная нормализация**

  
> **Минимакс** &mdash; линейное преобразование данных в диапазоне $[0, 1]$, где минимальное и максимальное масштабируемые значения соответствуют 0 и 1 соответственно.
> Реализуется оно по следующей формуле:
>
> $$\widetilde{x}_{ij} = \frac{x_{ij} - m_j}{M_j - m_j}$$
>
> где $x_{ij}$ &mdash; значение признака $j$ для объекта $i$, а $m_j$ и $M_j$ &mdash; минимальное и максимальное значения признака $j$ в *обучающем* наборе данных.  
> В модуле `sklearn.preprocessing` есть [класс](https://scikit-learn.org/stable/modules/generated/sklearn.preprocessing.MinMaxScaler.html) `MinMaxScaler`, который реализует эту операцию. Давайте рассмотрим методы этого класса подробнее:
>
> * `fit(X)`: вычисляет минимальные и максимальные значения для каждого признака.
>
> * `transform(X)`: применяет масштабирование к данным.
>
> * `fit_transform(X)`: объединяет оба метода, сначала подгоняя модель, а затем трансформируя данные.
>
> * `inverse_transform(X)`: восстанавливает исходные данные из масштабированных.
> 
> Обратите внимание, что методы `fit(X)` и `fit_transform(X)` можно применять только к обучающим данным.


Стоит отметить что существуют и другие методы масштабирования данных.

```python
#Преобразование в DataFrame
X_test_spd = pd.DataFrame(X_test_scaled,
                          columns=X_test.columns,
                          index=X_test.index)

X_train_spd = pd.DataFrame(X_train_scaled,
                           columns=X_train.columns,
                           index=X_train.index)

#Вывод максимумов и минимумов
print('Максимальные значения каждого признака')
print(X_train_spd.max())

print('\nМинимальные значения каждого признака:')
print(X_train_spd.min())
```

Обучите модель и сравните результат предсказаний со стандартизацией и без.

```python
print(f"Точность равна: {np.round(accuracy(y_test,
                                           knn(X_train_spd, y_train,
                                               X_test_spd, 5)) * 100,
                                  decimals=2)} %")
```

**Вывод:** Для knn нормировка данных сильно увеличивает предсказательную точность, т.к. они становится более равнозначными между собой

**А можем ли мы как-то еще повысить точность модели?**

Как мы видели, параметр $k$ числа соседей является необучаемым т.е. гиперпараметром, который выставляется вручную. Выбор оптимального значения $k$ важен, так как слишком маленькое значение может привести к переобучению, а слишком большое — к недообучению.



Постройте графики зависимостей точности (`accuracy_score`) от значения $k$ (от 1 до 20) для обучающей и тестовой выборок, на одной фигуре должно быть нарисовано две зависимости.

<font color="green" size="5"><b>Исправления ниже</b></font>


```python
test_accuracy = []
train_accuracy = []

#Подсчитываем accuracy для каждого k
for k in range(1, 21):
    train_accuracy.append(np.round(accuracy(y_train, knn(X_train_spd, y_train, X_train_spd, k)) * 100, decimals=2))
    test_accuracy.append(np.round(accuracy(y_test, knn(X_train_spd, y_train, X_test_spd, k)) * 100, decimals=2))
```

#### 3. Рестораны

Будем считать компанию рестораном, если в поле `categories` *содержится* слово `restaurant`. Обратите внимание, что в анализе данных часто нет четкого формата данных. Например, данное слово может быть написано как с большой буквы, так и с маленькой; может как разделяться `;`, так и не разделяться. При возникновении недопонимания стоит посмотреть данные.

Составьте таблицу, в которой будет информация о всех ресторанах города `N`, для которых имеется не менее $5$ отзывов. Далее постройте heatmap-график районов, в котором каждому району сопоставьте среднюю оценку по ресторанам этого района. Рассматривайте только те районы, в которых есть не менее $10$ ресторанов, для каждого из которых есть не менее $5$ отзывов.

```python
restaurant_mask = business_joined['categories'].str.contains('restaurant', case=False, na=False, regex=False)
restaurants = business_joined.loc[restaurant_mask].copy() # Создаем таблицу всех ресторанов

restaurant_pop = restaurants.loc[restaurants['number_of_reviews'] >= 5].copy() # Таблица всех популярных ресторанов

table_cnt_rest = pd.pivot_table(data=restaurant_pop,
                       index="lat_rounded",
                       columns=["lon_rounded"],
                       aggfunc="size") #Создание агрегированной таблицы (кол-во популярных ресторанов в районе)


plt.figure(figsize=(12,10))
ax_3 = sns.heatmap(table.mask(table_cnt_rest < 10), #Фильтрация районов с маленьким количеством ресторанов
                   cmap='RdYlGn',
                   vmin=1, vmax=5)
ax_3.set(xlabel="Долгота", ylabel="Широта",
         title='Средняя оценка популярных ресторанов по каждому району',
         xticks=[], yticks=[]) #Создание и настройка графика

plt.show()
```
