### Задача 1.

*Для начала простая задача для разминки.*

Рассмотрим модель одномерной линейной регрессии *без свободного параметра* $y(x) = \theta x$, где $x\in \mathbb{R}$ и $\theta \in \mathbb{R}$. Предполагаем, что данные получены по правилу
$$Y_i = \theta x_i + \varepsilon_i,$$
где $i \in \{1, ..., n\}$, числа $x_i \in \mathbb{R}$ неслучайны, а $\varepsilon_i$ &mdash; случайные ошибки.

**1.** Посчитайте оценку $\widehat{\theta}$ методом наименьших квадратов по явной формуле. В ответе во всех пунктах могут оставаться суммы $n$ слагаемых, но не должно быть операций с векторами или матрицами.

<font color="green" size="5"><b>Исправления ниже</b></font>


Запишем для выражения суммы квадратов отклонений:

$$
L(\theta)=\sum_{i=1}^{n}\bigl(Y_i-\theta\,x_i\bigr)^2.
$$

Это выражение — многочлен второй степени по $\theta$. Найдём производные:

$$
L'(\theta)=\sum_{i=1}^{n}\frac{d}{d\theta}\bigl(Y_i-\theta\,x_i\bigr)^2
=\sum_{i=1}^{n}2\bigl(Y_i-\theta\,x_i\bigr)(-x_i)
=\sum_{i=1}^{n}\left(2\theta\,x_i^2-2Y_ix_i\right).
$$

$$
L''(\theta)=\sum_{i=1}^{n}2x_i^2>0.
$$

Значит, $L(\theta)$ достигает минимума в точке, где $L'(\theta)=0$:

$$
0=\sum_{i=1}^{n}\left(2\hat\theta\,x_i^2-2Y_ix_i\right).
$$

Отсюда:

$$
\hat\theta\,\sum_{i=1}^{n}x_i^2=\sum_{i=1}^{n}Y_ix_i,
\qquad
\hat\theta=\frac{\sum_{i=1}^{n}Y_ix_i}{\sum_{i=1}^{n}x_i^2}.
$$

<font color="green" size="5"><b>Исправления выше</b></font>


**2.** Выпишите формулы итераций градиентного спуска (GD) и стохастического градиентного спуска (SGD) для поиска $\widehat{\theta}$. Не забудьте указать, как в SGD математически определяется батч (набор объектов на каждой итерации).

<font color="green" size="5"><b>Исправления ниже</b></font>


$$
\nabla L(\theta)=\sum_{i=1}^{n}\frac{d}{d\theta}\bigl(Y_i-\theta\,x_i\bigr)^2
=\sum_{i=1}^{n}2\bigl(Y_i-\theta\,x_i\bigr)(-x_i)
$$

$$
\nabla L(\theta)=2(\theta (\sum_{i=1}^{n}x_i^2 - \sum_{i=1}^{n}Y_i x_i)
$$

Шаг градиентного спуска:

$$
\theta_{t+1}
= \theta_t
- \eta \nabla L(\theta_t)
$$


Занесем 2 под $\eta$

$$
\theta_{t+1}
= \theta_t
- \eta (\theta_t \sum_{i=1}^{n}x_i^2 - \sum_{i=1}^{n}Y_i x_i)
$$


Шаг стохастического градиентного спуска:


$$
\theta_{t+1}
= \theta_t
- \eta (\theta_t \sum_{i \in Batch_t}x_i^2 - \sum_{i \in Batch_t}Y_i x_i)
$$


в $Batch_t$ это последовательность индексов из 1, 2, ..., n,
длины k, где каждый индекс выбирается независимо и равновероятно

<font color="green" size="5"><b>Исправления выше</b></font>


**3.** Опишите, что произойдет, если для некоторого $i$ ошибка $\varepsilon_i$ будет очень большой.

<font color="green" size="5"><b>Исправления ниже</b></font>


$$
\hat\theta=\frac{\sum_{i=1}^{n}Y_ix_i}{\sum_{i=1}^{n}x_i^2}.
$$


$$
\hat\theta=\frac{\sum_{i=1}^{n}(\theta x_i + \varepsilon_i)x_i}{\sum_{i=1}^{n}x_i^2}= \theta + \frac{\sum_{i=1}^{n} \varepsilon_i x_i}{\sum_{i=1}^{n}x_i^2}
$$

То есть зависимость линейная от каждого из $\varepsilon_i$, то есть если устремить его в бесконечность, то коэфиценты тоже устремятся в бесконечность

Это значение будет выбросом и будет сильно влиять на метрику и итоговые коэфиценты, т.к. зависимость от $\varepsilon_i$ квадратичная

*Теперь давайте углубимся в теорию градиентных методов оптимизации. Эти подходы будут часто встречаться, поэтому важно освоить их применение. Для этого применим градиентные методы в другом методе обучения линейной регрессии.*

В этой задаче вам предлагается реализовать регрессию Хьюбера, а также применить ее к данным с выбросами. Для начала реализуйте класс по шаблону снизу. Обратите внимание, что класс `HuberRegression` &mdash; наследник класса `BaseEstimator`, это с легкостью позволит использовать наш класс в различных пайплайнах библиотеки `sklearn`.

**1.** Рассмотрим линейную регрессию $y(x) = x^T \theta$, причем для оценки $\theta$ будем рассматривать функцию потерь Хьюбера
$$R(x) = \frac{x^2}{2} I\{|x| \leqslant c\} + c\left(|x| - \frac{c}{2}\right)I\{|x| > c\}.$$

Тем самым задача оптимизации имеет вид
$$\sum_{i=1}^n R(Y_i - x_i^T \theta) \longrightarrow \min_{\theta \in \mathbb{R}^d}.$$

Нарисуйте график $R(x)$

```python
def r_func(x, c=5):
    if abs(x) <= 5:
        return x**2 / 2
    else:
        return c * (abs(x) - c / 2)

plt.figure()
x = np.linspace(0, 10, 100)
y = np.vectorize(r_func)(x)
plt.plot(x, y)
plt.title('Функция потерь Хьюбера')
plt.xlabel('x')
plt.ylabel('y')

plt.tight_layout()
```

В чем польза выбора такой функции потерь?

<br/><details>
<summary> ➡️ Кликни для показа подсказки </summary>
    
1. Подумайте, что происходит в случае больших отклонений предсказаний от истины? Что это за точки?
    
2. Обратите внимание на поведение $R(x)$ в точке $x=c$.
</details>

Плюс этой функции заключается в том, что при больших отклонениях зависимость больше не квадратичная, а линейная, т.е. эта метрика меньше подвержена выбросам. Во вторых функция гладкая, т.е. в точке `x = c` производная непрерывная

Выпишите математические формулы градиентного спуска (GD) и стохастического градиентного спуска (SGD).

<font color="green" size="5"><b>Исправления ниже</b></font>


$$
\nabla \sum_{i=1}^{n} R\!\left(y_i - x_i^{\top}\theta\right)
= \sum_{i=1}^{n} \nabla R\!\left(y_i - x_i^{\top}\theta\right)
= \sum_{i=1}^{n}\Bigl(
- x_i\,(y_i - x_i^{\top}\theta)\,\mathbb{I}\{|x|\le c\}
- x_i\,\operatorname{sign}(y_i - x_i^{\top}\theta)\,\mathbb{I}\{|x|>c\}
\Bigr).
$$
Шаг градиентного спуска:
$$
\theta_{t+1}
= \theta_t
+ \eta \sum_{i=1}^{n} x_i\Bigl(
\mathbb{I}\{|x|\le c\}\,(y_i - x_i^{\top}\theta)
+ \mathbb{I}\{|x|>c\}\,\operatorname{sign}(y_i - x_i^{\top}\theta)
\Bigr).
$$

Аналогично для шага стохастического градиентного градиентного спуска:
$$
\theta_{t+1}
= \theta_t
+ \eta \sum_{i\in Batch_t} x_i\Bigl(
\mathbb{I}\{|x|\le c\}\,(y_i - x_i^{\top}\theta)
+ \mathbb{I}\{|x|>c\}\,\operatorname{sign}(y_i - x_i^{\top}\theta)
\Bigr).
$$


$Batch_t$ это последовательность индексов из 1, 2, ..., n, длины k, где каждый индекс выбирается независимо и равновероятно

<font color="green" size="5"><b>Исправления выше</b></font>

**2.** Реализуем теперь класс. **При реализации класса запрещено пользоваться ИИ-инструментами.**

<font color="green" size="5"><b>Исправления ниже</b></font>


```python
# При реализации класса запрещено пользоваться ИИ-инструментами.

BATCH_SIZE = 32


class HuberRegression(BaseEstimator):
    """Класс, реализующий линейную регрессию с функцией потерь Хьюбера."""

    def __init__(self, c: float = 1.0, fit_intercept: bool = True,
                 max_iter: int = 1000, step: float = 1.0, botomn: float = 0.0001) -> None:
        """Инициализирует модель.

        Параметры: c (float): Константа из функции потерь Хьюбера.
        step (float): Шаг градиентного спуска.
        botomn (float): Значение, при котором градиентный спуск останавливается.
        fit_intercept (bool): Добавлять ли константный признак. max_iter
        (int): Максимальное число итераций оптимизации.
        """
        self.c = c
        self.fit_intercept = fit_intercept
        self.max_iter = max_iter
        self.step = step
        self.botomn = botomn

    def fit(self, X: np.ndarray, y: np.ndarray) -> "HuberRegression":
        """Обучает модель.

        Параметры:
        X (np.ndarray): Матрица признаков.
        y (np.ndarray): Вектор целевой переменной.

        Возвращает:
        HuberRegression: Обученная модель.
        """


        X = np.asarray(X)
        y = np.asarray(y)

        if X.shape[0] != y.shape[0]:
            raise ValueError("Количество строк в X и y должно совпадать")

        # Добавление свободного члена
        if self.fit_intercept:
            X = np.hstack((X, np.ones((X.shape[0], 1))))

        self.coef_ = np.zeros(X.shape[1])
        self.n_iter_ = 0

        # Итерации градиентного спуска
        for k in range(self.max_iter):

            # Выбор batch
            batch_idx = np.random.choice(X.shape[0], size=BATCH_SIZE, replace=True)
            X_batch = X[batch_idx]
            y_batch = y[batch_idx]


        
            r = y_batch - X_batch @ self.coef_
            grad = np.zeros(X_batch.shape[1])

            # Вычисление градианта
            for i in range(BATCH_SIZE):
                if abs(r[i]) <= self.c:
                    grad += r[i] * X_batch[i]
                else:
                    grad += self.c * np.sign(r[i]) * X_batch[i]
            delta = grad * self.step / BATCH_SIZE
            
            # Проверка на пороговое значение
            if np.linalg.norm(delta) < self.botomn:
                break
            else:
                self.n_iter_ += 1
                self.coef_ += delta

        # Добавление свободного члена в поле класса
        self.intercept_ = 0
        if self.fit_intercept:
            self.intercept_ = self.coef_[-1]
            self.coef_ = self.coef_[:-1]

        return self

    def predict(self, X: np.ndarray) -> np.ndarray:
        """Делает предсказание на новых данных.

        Параметры: X (np.ndarray): Матрица признаков.

        Возвращает: np.ndarray: Вектор предсказанных значений.
        """

        if X.shape[1] != self.coef_.shape[0]:
            raise ValueError("Число признаков в X не соответствует числу коэффициентов модели")

        pred = X @ self.coef_ + self.intercept_
        return pred
```

### Задача 4.

*Наконец, давайте исследуем еще один метод обучения линейных моделей. Эта задача теоретическая, кодить ничего не требуется. Мы рассмотрим только случай простой модели, потому как в более сложном случае решение существенно усложняется.*


Рассмотрим модель одномерной регрессии $y(x) = \theta x$, где $x \in \mathbb{R}$ &mdash; одномерный признак, $y \in \mathbb{R}$ &mdash; целевой признак, $\theta \in \mathbb{R}$ &mdash; неизвестный параметр. Имеется выборка размера $n$, полученная по правилу
	$$Y_i = \theta x_i + \varepsilon_i,\ \ \ i=1,...,n,$$
где $\varepsilon_i$ &mdash; случайная ошибка измерений.

Оценку параметра $\theta$ будем искать методом наименьших модулей, то есть $$\sum_{i=1}^n \left|Y_i - \theta x_i\right| \to \min_\theta.$$

Какие преимущества и недостатки такой модели по сравнению с методом наименьших квадратов?

**Плюсы**
* Из-за того, что ошибка не возводится в квадрат, то метрика более стабильна к сильным выбросам
* Более понятна, т.к. ищет центр масс (медиану), т.е. мы просто минимизируем суммарное отклонение

**Минусы**
* Из-за того, что функция не гладкая, то ее сложнее оптимизировать
* Сложнее теоритически анализировать по аналогичной причине
* Может иметь много решений (опять же из-за того что может быть участок с нулевой производной) и из-за этого может появляться неоднозначность и ухудшение точности

Предложите точный алгоритм поиска оценки параметра $\theta$, работающий за время $O(n \log n)$. Приведите его описание и теоретическое обоснование.

Во-первых, заметим, что функция потерь при раскрытии модулей является непрерывной кусочно-линейной функцией. Из этого получаем, что минимум находится в точках перелома.

Точки перелома можем найти как нули соответствующих модулей из суммы:
$$
\theta_i = \frac{Y_i}{x_i}.
$$
Случаи с $x_i = 0$ выкидываем, так как $\theta$ в соответствующих модулях не влияет.

Тогда получаем, что
$$
L(\theta) = \sum_{i=1}^{n} |x_i|\,|\theta_i - \theta|.
$$

Отсортируем $\theta_i$ по возрастанию за $O(n \log n)$.

Заметим, что угол наклона при $\theta_k \le \theta \le \theta_{k+1}$ равен:
$$
\sum_{i=1}^{k} |x_i| - \sum_{i=k+1}^{n} |x_i|.
$$
Причём при $\theta < \theta_1$ он меньше нуля, а при $\theta > \theta_n$ — больше.

Значит, минимум функции $L$ среди $\theta_i$.

Пройдёмся по $\theta_i$ от $1$ до $n$. Находясь на значении $\theta_k$, храним в памяти коэффициент наклона:
$$
\mathrm{Coef}(\theta_k) = \sum_{i=1}^{k} |x_i| - \sum_{i=k+1}^{n} |x_i|.
$$
Также храним текущий минимум потерь и соответствующий ему $\theta$.

Функция кусочно-линейная, значит
$$
L(\theta_{k+1}) = L(\theta_k) + (\theta_{k+1} - \theta_k)\,\mathrm{Coef}(\theta_k).
$$

Тогда получаем:
$$
\mathrm{Coef}(\theta_{k+1}) = \mathrm{Coef}(\theta_k) + 2|x_k|.
$$

Оба пересчёта делаются за $O(1)$, тем самым мы проходим по всем $n$ значениям $\theta$ за $O(n)$.

Также нужно сделать подсчёт $\mathrm{Coef}(\theta_1)$ и $L(\theta_1)$ за $O(n)$.

Итоговая сложность: $O(n \log n)$.
